```python
import pandas as pd

# Load 2022 validation data and prescription data
df_2022 = pd.read_csv('../dataset/data_2022.csv')
rx_2022 = pd.read_csv('../grouped_dataset/prescription_2022.csv')

if 'Unnamed: 0' in df_2022.columns:
    df_2022 = df_2022.drop(columns=['Unnamed: 0'])
if 'Unnamed: 0' in rx_2022.columns:
    rx_2022 = rx_2022.drop(columns=['Unnamed: 0'])

print("2022 integrated shape:", df_2022.shape)
print("2022 prescription shape:", rx_2022.shape)
print("\n2022 integrated columns:", df_2022.columns.tolist())
print("2022 prescription columns:", rx_2022.columns.tolist())
```

```python
# Select only suggested prescription features
rx_2022_selected = rx_2022[['Observation_ID', 'Quantity', 'Form', 'Strength', 'Day_Supply']].copy()

# Join onto 2022 integrated data
super_2022 = df_2022.merge(rx_2022_selected, on='Observation_ID', how='left')

print("Super 2022 shape:", super_2022.shape)
print("Columns:", super_2022.columns.tolist())
print("\nMissing values:")
print(super_2022.isnull().sum())
```

```python
# Check if Observation_IDs overlap between 2022 integrated and 2022 prescription
integrated_ids = set(df_2022['Observation_ID'].unique())
prescription_ids = set(rx_2022['Observation_ID'].unique())

print(f"Observation_IDs in 2022 integrated:   {len(integrated_ids)}")
print(f"Observation_IDs in 2022 prescription: {len(prescription_ids)}")
print(f"Overlap:                              {len(integrated_ids & prescription_ids)}")
print(f"\nSample IDs from integrated:   {list(integrated_ids)[:5]}")
print(f"Sample IDs from prescription: {list(prescription_ids)[:5]}")
```

```python
# Select only  suggested prescription features
rx_2022_selected = rx_2022[['Observation_ID', 'Quantity', 'Form', 'Strength', 'Day_Supply']].copy()

# Join onto 2022 integrated data
super_2022 = df_2022.merge(rx_2022_selected, on='Observation_ID', how='left')

print("Super 2022 shape:", super_2022.shape)
print("Columns:", super_2022.columns.tolist())
print("\nMissing values:")
print(super_2022.isnull().sum())

# Save
super_2022.to_csv('super_data_2022.csv', index=False)
print("\nSaved: super_data_2022.csv")
```
